```python
import geopandas as gpd
miasta = gpd.read_file("dane/pl_miasta_wojewódzkie.geojson")
pasy = gpd.read_file("dane/pl_pasy_ukształtowania.geojson")
rzeki = gpd.read_file("dane/pl_rzeki.geojson")
```

```python
#ZADANIE A)

pasy.to_crs("EPSG:2180", inplace=True)
#print(pasy.crs)
print(pasy.geometry.area)
```

```python
#ZADANIE B)

miasta.to_crs("EPSG:2180", inplace=True)
bb = miasta.geometry.total_bounds
print(((bb[2] - bb[0])*2 + (bb[3] - bb[1])*2)/1000, "km")
```
