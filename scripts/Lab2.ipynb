```python
import pandas as pd
from statistics import mean
import numpy as np
import os, json
from random import sample
```

```python
#ZADANIE A)

data = {
    "name": ["Alicja", "Bartłomiej", "Celina", "Damian", "Eryk", "Filip", "Grzegorz", "Hubert", "Imian", "Jolanta"],
    "height": sample(range(140,190), 10),
    "weight": sample(range(45,100), 10)
}
#print(data)
df = pd.DataFrame(data)

df["bmi"] = round(df["weight"]/(df["height"]**2)*10000, 2)

print(df)
```

```python
#ZADANIE B)

data2 = {
    "name": ["Rysy", "Krywań", "Mont Buet"],
    "height": [2499, 2494, 3096],
    "country": ["Poland", "Slovakia", "France"]
}

df2 = pd.DataFrame(data2)

json1 = json.dumps(data2, indent=4)
with open("json1.json", "w") as j:
    j.write(json1)

with open("json1.json", "r") as f:
    df3 = pd.DataFrame(json.load(f))

print(df2.equals(df3))
```
