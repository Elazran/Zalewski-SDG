```python
from pathlib import Path
import tkinter as tk

# Definicje metod:

def close_program():
    window.destroy()


# Utworzenie głównego okna + canvas:

window = tk.Tk()
window.title("VectorTools")
window.resizable(False, False)

canvas = tk.Canvas(
    window,
    width=650,
    height=550,
    highlightthickness=0
)

canvas.pack()


# Grafika tła:

image_path = Path(__file__).resolve().with_name("MlkWay.png")
background_screen = tk.PhotoImage(file=str("C:\Users\filip\Zalewski-SDG\scripts\MlkWay.png"))

canvas.create_image(
    0,
    0,
    image=background_screen,
    anchor="nw"
)


# Tytuł programu:
canvas.create_text(
    200,
    50,
    text="Voyage El2",
    font=("Arial", 30, "bold"),
    fill="white"
)

# Menu:

canvas.create_text(
    180,
    420,
    text="Obierz kurs:",
    font=("Arial", 16, "bold"),
    fill="white"
)


# Przycisk zamknięcia:

canvas.create_text(
    180,
    465,
    text="Opuść pokład",
    font=("Arial", 11, "bold"),
    fill="white",
    tags="zamknij"
)

canvas.tag_bind(
    "zamknij",
    "<Button-1>",
    lambda event: close_program()
)

canvas.tag_bind(
    "zamknij",
    "<Enter>",
    lambda event: canvas.itemconfig(
        "zamknij",
        fill="#90E0EF"
    )
)

canvas.tag_bind(
    "zamknij",
    "<Leave>",
    lambda event: canvas.itemconfig(
        "zamknij",
        fill="white"
    )
)

# Główna pętla programu:

window.mainloop()
```
