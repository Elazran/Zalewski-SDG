```python
import geopandas as gpd
```

```python
rzeki = gpd.read_file("dane/hr1.geojson")
kraje = gpd.read_file("dane/lsp1.geojson")
```

```python
#ZADANIE a)

#print(len(rzeki.geometry.iloc[1].geoms))

liczba = rzeki.geometry.apply(
    lambda geom: 2 <= len(geom.geoms) <= 10)

print(liczba.sum())
```

```python
#ZADANIE b)

#print(kraje.head())

kraje["ludność_w_wieku_65_lat_i_więcej"] = (
    kraje["ludność_w_wieku_65_lat_i_więcej"].astype(str)
    .str.replace(" %", "", regex=False)
    .str.replace("brak danych", "0", regex=False)
    .str.replace(",", ".", regex=False))

kraje["ludność_w_wieku_65_lat_i_więcej"] = kraje["ludność_w_wieku_65_lat_i_więcej"].astype(float)

kraje65 = kraje[kraje["ludność_w_wieku_65_lat_i_więcej"] > 18]

kraje65.plot(column="ludność_w_wieku_65_lat_i_więcej", legend=True)
```
